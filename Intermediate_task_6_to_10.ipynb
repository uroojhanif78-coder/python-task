```python
# task : 6
total_products = int(input("enter total number of products:"))

total_bill=0
for i in range(total_products):
  print("\nproduct", i+1)
  name= input("enter product name:")
  price= float(input("enter price:"))
  qty=int(input("enter quantity:"))

  subtotal = price*qty
  print(f"{name} subtotal: {subtotal}")
  total_bill = total_bill + subtotal

  discount = 0
  if total_bill >= 5000:
    discount = total_bill*0.10
    print(f"discount applied: {discount}")

    final_bill = total_bill - discount
    print("\n====== FINAL BILL ======")
    print(f"total before discount: {total_bill}")
    print(f"discount: {discount}")
    print(f"amount to pay:{final_bill}")
```

```python
# task : 9 Message analyzer

message =  input("enter a message")

total_characters = len(message)

total_words = len(message.split())
vowels = "aeiouAEIOU"
total_vowels = 0

for char in message:
  if char in vowels:
    total_vowels =  total_vowels +1
    print("total vowels:" , total_vowels)

total_spaces = message.count(" ")

uppercase_message = message.upper()

python_exists = "Python" in message


print("\n--- Message Information---")
print("Total Characters:", total_characters)
print("Total Words:", total_words)
print("Total Spaces:", total_spaces)
print("Uppercase Message:", uppercase_message)
print("Does 'Python' exist?:", python_exists)
```

```python
# task 10
#Contact Book
contacts = {}

while True:
    print("\n1. Add Contact")
    print("2. Search Contact")
    print("3. Display All Contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter contact name: ")

        if name in contacts:
            print("Contact already exists!")
        else:
            phone = input("Enter phone number: ")
            contacts[name] = phone
            print("Contact added successfully!")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone Number:", contacts[name])
        else:
            print("Contact not found!")

    elif choice == "3":
        if len(contacts) == 0:
            print("No contacts available.")
        else:
            print("\nAll Contacts:")
            for name, phone in contacts.items():
                print(name, ":", phone)

    elif choice == "4":
        print("Program ended.")
        break

    else:
        print("Invalid choice!")
```
